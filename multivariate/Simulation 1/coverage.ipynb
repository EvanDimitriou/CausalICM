```python
import sys
import os
import pandas as pd
import numpy as np

# ---------------------------------------------------------------------
# Insert your *local path* to the CausalICM folder here
# Example: "/home/user/projects/CausalICM/"
# ---------------------------------------------------------------------
USER_PATH = "/path/to/your/CausalICM/folder/"
sys.path.append(USER_PATH)

# Import required functions
from CausalICM.functions.coverage_multivariate95 import coverage1

# ---------------------------------------------------------------------
# Load best rho value
# ---------------------------------------------------------------------
BEST_RHO_PATH = USER_PATH + "multivariate/Simulation1/sim1_best_rho.csv"
best_rho = float(pd.read_csv(BEST_RHO_PATH).values[0])

# ---------------------------------------------------------------------
# Run coverage experiment
# ---------------------------------------------------------------------
(
    mse_ICM, mse_GPobs, mse_GPexp,
    coverage_prob_ICM_95, coverage_prob_ICM_90,
    coverage_prob_GPobs_95, coverage_prob_GPobs_90,
    coverage_prob_GPexp_95, coverage_prob_GPexp_90,
    length_ICM90, length_ICM95,
    length_GPobs90, length_GPobs95,
    length_GPexp90, length_GPexp95
) = coverage1(
    num_datasets=200,
    sample_size=1000,
    beta0=-10,
    beta1=-8,
    beta2=-8,
    r=2,
    ID=5,
    AD=[0, 1, 2, 3, 4],
    rho=best_rho
)

# ---------------------------------------------------------------------
# Save results
# ---------------------------------------------------------------------
sim1_coverage_df = pd.DataFrame(
    np.c_[
        mse_ICM, mse_GPobs, mse_GPexp,
        coverage_prob_ICM_95, coverage_prob_ICM_90,
        coverage_prob_GPobs_95, coverage_prob_GPobs_90,
        coverage_prob_GPexp_95, coverage_prob_GPexp_90,
        length_ICM90, length_ICM95,
        length_GPobs90, length_GPobs95,
        length_GPexp90, length_GPexp95
    ],
    columns=[
        'mse_ICM', 'mse_GPobs', 'mse_GPexp',
        'coverage_prob_ICM_95', 'coverage_prob_ICM_90',
        'coverage_prob_GPobs_95', 'coverage_prob_GPobs_90',
        'coverage_prob_GPexp_95', 'coverage_prob_GPexp_90',
        'length_ICM90', 'length_ICM95',
        'length_GPobs90', 'length_GPobs95',
        'length_GPexp90', 'length_GPexp95'
    ]
)

sim1_coverage_df.to_csv("sim1_coverage.csv", index=False)
```
