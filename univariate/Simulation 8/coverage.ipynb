```python
import sys
import pandas as pd
import numpy as np

# Add local repo folder
sys.path.append("/Users/evangelosdimitriou/myGitRepo/multi-task-GPs-for-Treatment-Effect-Estimation-2/__pycache__/")
print(sys.path)

# Import coverage function
from neurIPS.functions.coverage import coverage8

# Read best rho
best_rho = pd.read_csv('sim8_best_rho.csv').values[0][0]

# Compute coverage, lengths, and MSEs for simulation 8
(mse_ICM, mse_GPobs, mse_GPexp, 
 coverage_prob_ICM_95, coverage_prob_ICM_90, 
 coverage_prob_GPobs_95, coverage_prob_GPobs_90, 
 coverage_prob_GPexp_95, coverage_prob_GPexp_90, 
 length_ICM90, length_ICM95, 
 length_GPobs90, length_GPobs95, 
 length_GPexp90, length_GPexp95) = coverage8(
    num_datasets=200,
    min=-2,
    max=2.02,
    step=0.02,
    sample_size=1000, 
    beta0=-3,
    beta1=-3,
    r=2,
    ID=1,
    AD=0,
    rho=best_rho
)

# -------------------------------
# Convert results to DataFrames
# -------------------------------
x_values = np.arange(-2, 2.02, 0.02)

sim8_coverage_df = pd.DataFrame(
    np.c_[x_values, coverage_prob_ICM_95, coverage_prob_ICM_90,
          coverage_prob_GPobs_95, coverage_prob_GPobs_90,
          coverage_prob_GPexp_95, coverage_prob_GPexp_90],
    columns=['test_data', 'coverageProbICM95', 'coverageProbICM90',
             'coverageProbGPobs95', 'coverageProbGPobs90',
             'coverageProbGPexp95', 'coverageProbGPexp90']
)

sim8_length_df = pd.DataFrame(
    np.c_[x_values, length_ICM95, length_ICM90,
          length_GPobs95, length_GPobs90,
          length_GPexp95, length_GPexp90],
    columns=['test_data', 'length_ICM95', 'length_ICM90',
             'length_GPobs95', 'length_GPobs90',
             'length_GPexp95', 'length_GPexp90']
)

sim8_mse_df = pd.DataFrame(
    np.c_[mse_ICM, mse_GPobs, mse_GPexp],
    columns=['mseICM', 'mseGPobs', 'mseGPexp']
)

# -------------------------------
# Save results to CSV
# -------------------------------
sim8_coverage_df.to_csv('sim8_coverage.csv', index=False)
sim8_length_df.to_csv('sim8_length.csv', index=False)
sim8_mse_df.to_csv('sim8_mse.csv', index=False)
```
