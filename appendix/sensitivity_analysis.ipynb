```python
# The effect of $\rho$
import numpy as np
from sklearn.metrics import mean_squared_error
import GPy
import sys
import pandas as pd
import matplotlib.pyplot as plt
sys.path.append("/path/to/your/CausalICM/")
print(sys.path)
from neurIPS.functions.data_simulation import data_simulation8
from neurIPS.functions.model_training import ICM
test_data = np.round(np.arange(-2.0, 2.04, 0.04), 2)
trueCATE = 1+test_data+test_data**2
datasets = pd.read_csv('/path/to/your/CausalICM/Simulation 8/datasets_sim8.csv')

## $\rho=0.0$
rho = 0.0

rmse_rho00 = np.zeros((100))
bias_rho00 = np.zeros((100))
var_rho00 = np.zeros((100))


for i in range(100):
    print(i + 1, end=" ")
    data_exp = datasets[(datasets['S'] == 1) & (datasets['datasetNo'] == i + 1)]
    data_obs = datasets[(datasets['S'] == 0) & (datasets['datasetNo'] == i + 1)]


    m0, m1 = ICM(X_E = data_exp["X"], X_O = data_obs["X"], 
                 T_E = data_exp["A"], T_O = data_obs["A"], 
                 Y_E = data_exp["Y"], Y_O = data_obs["Y"],  
                 r = 2, ID=1, AD=0, rho=rho)
    predY0 = m0.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predY1 = m1.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predCATE = predY1[0] - predY0[0]

    

    rmse_rho00[i] = np.sqrt(mean_squared_error((predCATE), (trueCATE)))
    bias_rho00[i] = np.mean(predCATE - trueCATE)
    var_rho00[i] = np.mean((predCATE - np.mean(predCATE))**2)
## $\rho=0.1$
rho = 0.1

rmse_rho01 = np.zeros((100))
bias_rho01 = np.zeros((100))
var_rho01 = np.zeros((100))

for i in range(100):
    print(i + 1, end=" ")
    data_exp = datasets[(datasets['S'] == 1) & (datasets['datasetNo'] == i + 1)]
    data_obs = datasets[(datasets['S'] == 0) & (datasets['datasetNo'] == i + 1)]


    m0, m1 = ICM(X_E = data_exp["X"], X_O = data_obs["X"], 
                 T_E = data_exp["A"], T_O = data_obs["A"], 
                 Y_E = data_exp["Y"], Y_O = data_obs["Y"],  
                 r = 2, ID=1, AD=0, rho=rho)
    predY0 = m0.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predY1 = m1.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predCATE = predY1[0] - predY0[0]

    

    rmse_rho01[i] = np.sqrt(mean_squared_error((predCATE), (trueCATE)))
    bias_rho01[i] = np.mean(predCATE - trueCATE)
    var_rho01[i] = np.mean((predCATE - np.mean(predCATE))**2)
## $\rho=0.2$
rho = 0.2

rmse_rho02 = np.zeros((100))
bias_rho02 = np.zeros((100))
var_rho02 = np.zeros((100))

for i in range(100):
    print(i + 1, end=" ")
    data_exp = datasets[(datasets['S'] == 1) & (datasets['datasetNo'] == i + 1)]
    data_obs = datasets[(datasets['S'] == 0) & (datasets['datasetNo'] == i + 1)]


    m0, m1 = ICM(X_E = data_exp["X"], X_O = data_obs["X"], 
                 T_E = data_exp["A"], T_O = data_obs["A"], 
                 Y_E = data_exp["Y"], Y_O = data_obs["Y"],  
                 r = 2, ID=1, AD=0, rho=rho)
    predY0 = m0.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predY1 = m1.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predCATE = predY1[0] - predY0[0]

    

    rmse_rho02[i] = np.sqrt(mean_squared_error((predCATE), (trueCATE)))
    bias_rho02[i] = np.mean(predCATE - trueCATE)
    var_rho02[i] = np.mean((predCATE - np.mean(predCATE))**2)
## $\rho=0.3$
rho = 0.3

rmse_rho03 = np.zeros((100))
bias_rho03 = np.zeros((100))
var_rho03 = np.zeros((100))

for i in range(100):
    print(i + 1, end=" ")
    data_exp = datasets[(datasets['S'] == 1) & (datasets['datasetNo'] == i + 1)]
    data_obs = datasets[(datasets['S'] == 0) & (datasets['datasetNo'] == i + 1)]


    m0, m1 = ICM(X_E = data_exp["X"], X_O = data_obs["X"], 
                 T_E = data_exp["A"], T_O = data_obs["A"], 
                 Y_E = data_exp["Y"], Y_O = data_obs["Y"],  
                 r = 2, ID=1, AD=0, rho=rho)
    predY0 = m0.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predY1 = m1.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predCATE = predY1[0] - predY0[0]

    

    rmse_rho03[i] = np.sqrt(mean_squared_error((predCATE), (trueCATE)))
    bias_rho03[i] = np.mean(predCATE - trueCATE)
    var_rho03[i] = np.mean((predCATE - np.mean(predCATE))**2)
## $\rho=0.4$
rho = 0.4

rmse_rho04 = np.zeros((100))
bias_rho04 = np.zeros((100))
var_rho04 = np.zeros((100))

for i in range(100):
    print(i + 1, end=" ")
    data_exp = datasets[(datasets['S'] == 1) & (datasets['datasetNo'] == i + 1)]
    data_obs = datasets[(datasets['S'] == 0) & (datasets['datasetNo'] == i + 1)]


    m0, m1 = ICM(X_E = data_exp["X"], X_O = data_obs["X"], 
                 T_E = data_exp["A"], T_O = data_obs["A"], 
                 Y_E = data_exp["Y"], Y_O = data_obs["Y"],  
                 r = 2, ID=1, AD=0, rho=rho)
    predY0 = m0.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predY1 = m1.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predCATE = predY1[0] - predY0[0]

    

    rmse_rho04[i] = np.sqrt(mean_squared_error((predCATE), (trueCATE)))
    bias_rho04[i] = np.mean(predCATE - trueCATE)
    var_rho04[i] = np.mean((predCATE - np.mean(predCATE))**2)
## $\rho=0.5$
rho = 0.5

rmse_rho05 = np.zeros((100))
bias_rho05 = np.zeros((100))
var_rho05 = np.zeros((100))

for i in range(100):
    print(i + 1, end=" ")
    data_exp = datasets[(datasets['S'] == 1) & (datasets['datasetNo'] == i + 1)]
    data_obs = datasets[(datasets['S'] == 0) & (datasets['datasetNo'] == i + 1)]


    m0, m1 = ICM(X_E = data_exp["X"], X_O = data_obs["X"], 
                 T_E = data_exp["A"], T_O = data_obs["A"], 
                 Y_E = data_exp["Y"], Y_O = data_obs["Y"],  
                 r = 2, ID=1, AD=0, rho=rho)
    predY0 = m0.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predY1 = m1.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predCATE = predY1[0] - predY0[0]

    

    rmse_rho05[i] = np.sqrt(mean_squared_error((predCATE), (trueCATE)))
    bias_rho05[i] = np.mean(predCATE - trueCATE)
    var_rho05[i] = np.mean((predCATE - np.mean(predCATE))**2)
## $\rho=0.6$
rho = 0.6

rmse_rho06 = np.zeros((100))
bias_rho06 = np.zeros((100))
var_rho06 = np.zeros((100))

for i in range(100):
    print(i + 1, end=" ")
    data_exp = datasets[(datasets['S'] == 1) & (datasets['datasetNo'] == i + 1)]
    data_obs = datasets[(datasets['S'] == 0) & (datasets['datasetNo'] == i + 1)]


    m0, m1 = ICM(X_E = data_exp["X"], X_O = data_obs["X"], 
                 T_E = data_exp["A"], T_O = data_obs["A"], 
                 Y_E = data_exp["Y"], Y_O = data_obs["Y"],  
                 r = 2, ID=1, AD=0, rho=rho)
    predY0 = m0.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predY1 = m1.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predCATE = predY1[0] - predY0[0]

    

    rmse_rho06[i] = np.sqrt(mean_squared_error((predCATE), (trueCATE)))
    bias_rho06[i] = np.mean(predCATE - trueCATE)
    var_rho06[i] = np.mean((predCATE - np.mean(predCATE))**2)
## $\rho=0.7$
rho = 0.7

rmse_rho07 = np.zeros((100))
bias_rho07 = np.zeros((100))
var_rho07 = np.zeros((100))

for i in range(100):
    print(i + 1, end=" ")
    data_exp = datasets[(datasets['S'] == 1) & (datasets['datasetNo'] == i + 1)]
    data_obs = datasets[(datasets['S'] == 0) & (datasets['datasetNo'] == i + 1)]


    m0, m1 = ICM(X_E = data_exp["X"], X_O = data_obs["X"], 
                 T_E = data_exp["A"], T_O = data_obs["A"], 
                 Y_E = data_exp["Y"], Y_O = data_obs["Y"],  
                 r = 2, ID=1, AD=0, rho=rho)
    predY0 = m0.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predY1 = m1.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predCATE = predY1[0] - predY0[0]

    

    rmse_rho07[i] = np.sqrt(mean_squared_error((predCATE), (trueCATE)))
    bias_rho07[i] = np.mean(predCATE - trueCATE)
    var_rho07[i] = np.mean((predCATE - np.mean(predCATE))**2)
## $\rho=0.8$
rho = 0.8

rmse_rho08 = np.zeros((100))
bias_rho08 = np.zeros((100))
var_rho08 = np.zeros((100))

for i in range(100):
    print(i + 1, end=" ")
    data_exp = datasets[(datasets['S'] == 1) & (datasets['datasetNo'] == i + 1)]
    data_obs = datasets[(datasets['S'] == 0) & (datasets['datasetNo'] == i + 1)]


    m0, m1 = ICM(X_E = data_exp["X"], X_O = data_obs["X"], 
                 T_E = data_exp["A"], T_O = data_obs["A"], 
                 Y_E = data_exp["Y"], Y_O = data_obs["Y"],  
                 r = 2, ID=1, AD=0, rho=rho)
    predY0 = m0.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predY1 = m1.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predCATE = predY1[0] - predY0[0]

    

    rmse_rho08[i] = np.sqrt(mean_squared_error((predCATE), (trueCATE)))
    bias_rho08[i] = np.mean(predCATE - trueCATE)
    var_rho08[i] = np.mean((predCATE - np.mean(predCATE))**2)
## $\rho=0.9$
rho = 0.9

rmse_rho09 = np.zeros((100))
bias_rho09 = np.zeros((100))
var_rho09 = np.zeros((100))

for i in range(100):
    print(i + 1, end=" ")
    data_exp = datasets[(datasets['S'] == 1) & (datasets['datasetNo'] == i + 1)]
    data_obs = datasets[(datasets['S'] == 0) & (datasets['datasetNo'] == i + 1)]


    m0, m1 = ICM(X_E = data_exp["X"], X_O = data_obs["X"], 
                 T_E = data_exp["A"], T_O = data_obs["A"], 
                 Y_E = data_exp["Y"], Y_O = data_obs["Y"],  
                 r = 2, ID=1, AD=0, rho=rho)
    predY0 = m0.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predY1 = m1.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predCATE = predY1[0] - predY0[0]

    

    rmse_rho09[i] = np.sqrt(mean_squared_error((predCATE), (trueCATE)))
    bias_rho09[i] = np.mean(predCATE - trueCATE)
    var_rho09[i] = np.mean((predCATE - np.mean(predCATE))**2)
## $\rho=1.0$
rho = 1.0

rmse_rho10 = np.zeros((100))
bias_rho10 = np.zeros((100))
var_rho10 = np.zeros((100))

for i in range(100):
    print(i + 1, end=" ")
    data_exp = datasets[(datasets['S'] == 1) & (datasets['datasetNo'] == i + 1)]
    data_obs = datasets[(datasets['S'] == 0) & (datasets['datasetNo'] == i + 1)]


    m0, m1 = ICM(X_E = data_exp["X"], X_O = data_obs["X"], 
                 T_E = data_exp["A"], T_O = data_obs["A"], 
                 Y_E = data_exp["Y"], Y_O = data_obs["Y"],  
                 r = 2, ID=1, AD=0, rho=rho)
    predY0 = m0.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predY1 = m1.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predCATE = predY1[0] - predY0[0]

    

    rmse_rho10[i] = np.sqrt(mean_squared_error((predCATE), (trueCATE)))
    bias_rho10[i] = np.mean(predCATE - trueCATE)
    var_rho10[i] = np.mean((predCATE - np.mean(predCATE))**2)
print(np.round(np.mean(rmse_rho00), 3))
print(np.round(np.std(rmse_rho00), 3))
print('\n')
print(np.round(np.mean(rmse_rho01), 3))
print(np.round(np.std(rmse_rho01), 3))
print('\n')
print(np.round(np.mean(rmse_rho02), 3))
print(np.round(np.std(rmse_rho02), 3))
print('\n')
print(np.round(np.mean(rmse_rho03), 3))
print(np.round(np.std(rmse_rho03), 3))
print('\n')
print(np.round(np.mean(rmse_rho04), 3))
print(np.round(np.std(rmse_rho04), 3))
print('\n')
print(np.round(np.mean(rmse_rho05), 3))
print(np.round(np.std(rmse_rho05), 3))
print('\n')
print(np.round(np.mean(rmse_rho06), 3))
print(np.round(np.std(rmse_rho06), 3))
print('\n')
print(np.round(np.mean(rmse_rho07), 3))
print(np.round(np.std(rmse_rho07), 3))
print('\n')
print(np.round(np.mean(rmse_rho08), 3))
print(np.round(np.std(rmse_rho08), 3))
print('\n')
print(np.round(np.mean(rmse_rho09), 3))
print(np.round(np.std(rmse_rho09), 3))
print('\n')
print(np.round(np.mean(rmse_rho10), 3))
print(np.round(np.std(rmse_rho10), 3))

# Save the results
sens_rho00 = pd.DataFrame({"RMSE": rmse_rho00, "Bias": bias_rho00, "Variance": var_rho00})
sens_rho01 = pd.DataFrame({"RMSE": rmse_rho01, "Bias": bias_rho01, "Variance": var_rho01})
sens_rho02 = pd.DataFrame({"RMSE": rmse_rho02, "Bias": bias_rho02, "Variance": var_rho02})
sens_rho03 = pd.DataFrame({"RMSE": rmse_rho03, "Bias": bias_rho03, "Variance": var_rho03})
sens_rho04 = pd.DataFrame({"RMSE": rmse_rho04, "Bias": bias_rho04, "Variance": var_rho04})
sens_rho05 = pd.DataFrame({"RMSE": rmse_rho05, "Bias": bias_rho05, "Variance": var_rho05})
sens_rho06 = pd.DataFrame({"RMSE": rmse_rho06, "Bias": bias_rho06, "Variance": var_rho06})
sens_rho07 = pd.DataFrame({"RMSE": rmse_rho07, "Bias": bias_rho07, "Variance": var_rho07})
sens_rho08 = pd.DataFrame({"RMSE": rmse_rho08, "Bias": bias_rho08, "Variance": var_rho08})
sens_rho09 = pd.DataFrame({"RMSE": rmse_rho09, "Bias": bias_rho09, "Variance": var_rho09})
sens_rho10 = pd.DataFrame({"RMSE": rmse_rho10, "Bias": bias_rho10, "Variance": var_rho10})

# Create a list of the DataFrames
dfs = [sens_rho00, sens_rho01, sens_rho02, sens_rho03, sens_rho04, 
       sens_rho05, sens_rho06, sens_rho07, sens_rho08, sens_rho09, sens_rho10]

# Create a list of rho values
rho_values = [i * 0.1 for i in range(11)]

# Add the 'rho' column to each DataFrame
for df, rho in zip(dfs, rho_values):
    df['rho'] = rho

# Concatenate all the DataFrames into one
sens_rho = pd.concat(dfs, ignore_index=True)

sens_rho.to_csv('sens_rho.csv', index=False)

# Display the combined DataFrame
print(sens_rho)

########################################################################


# Observational/Experimental Sample Size

# Data Generation Function

# Data generation
from scipy.stats import bernoulli, norm

def data_simulation(sample_size_exp, sample_size_obs):
    
    # RCT
    X_trial = np.random.uniform(-2, 0.5, sample_size_exp)
    S_trial = np.ones(sample_size_exp)
    A_trial = bernoulli.rvs(0.5, size=sample_size_exp)
    tau_X_trial = 1 + X_trial + X_trial**2
    epsilon_trial = norm.rvs(0, 1.0, sample_size_exp)
    Y_trial = A_trial * tau_X_trial + X_trial**2 - 1 + epsilon_trial

    # Observational study
    X_observational = np.random.uniform(-2, 2, sample_size_obs)
    S_observational = np.zeros(sample_size_obs)
    logit_e_X_0 = X_observational
    A_observational = bernoulli.rvs(1 / (1 + np.exp(-logit_e_X_0)), size=sample_size_obs)
    tau_X_observational = 1 + X_observational + X_observational**2
    U_observational = (2 * A_observational - 1) *np.sin(X_observational-1) + norm.rvs(0, 1, size=sample_size_obs)
    epsilon_observational = norm.rvs(0, 1, size=sample_size_obs)
    Y_observational = A_observational * tau_X_observational + X_observational**2 - 1 + U_observational + epsilon_observational

    # Create DataFrames
    data_trial = np.column_stack((S_trial, X_trial, A_trial, Y_trial))
    data_observational = np.column_stack((S_observational, X_observational, A_observational, Y_observational))

    return data_trial, data_observational


## $n_{obs} = n_{exp} = 250$
### Tune rho
from sklearn import linear_model
from sklearn.model_selection import KFold
from sklearn.metrics import mean_squared_error
import numpy.random as nprand
from neurIPS.functions.model_training import ICM
from neurIPS.functions.weightedMSE import weighted_mean_squared_error

data_exp, data_obs = data_simulation(sample_size_exp=250, sample_size_obs=250)
data_full = np.r_[data_exp, data_obs]

logr = linear_model.LogisticRegression()
logr.fit(data_full[:,1].reshape(-1,1),data_full[:,0])
prob = logr.predict_proba(data_full[:,1].reshape(-1,1))
prob_part = prob[:,1]
prob_nonpart = prob[:,0]
data_full = np.c_[data_full, 1/prob_part]

seed_value = 1234

# Define range of rho values to search over
rho_values = np.round(np.arange(0, 1.05, 0.1), 1)

# Define number of folds for cross-validation
num_folds = 5  # Example: 5-fold cross-validation

# Initialize variables to store results
cv_rmse_values = []
rho_rmse_values = {}  # Dictionary to store RMSE values for each rho
min_avg_wmse = float('inf')
best_rho_nobs250 = None
wmse_per_rho = {}

# Perform 5-fold cross-validation
kf = KFold(n_splits=num_folds, shuffle=True)
for rho in rho_values:
    
    print(f"Rho: {rho}")
    avg_wmse = 0
    fold_num = 0
    wmse_per_fold = []
    
    for train_index, val_index in kf.split(data_full):
        # 5 fold CV for experimental data: split them into 5 folds
        train_data, val_data= data_full[train_index], data_full[val_index]
        val_data = val_data[val_data[:,0]==1,:]
        X_val=val_data[:,1]
        T_val=val_data[:,2]
        Y_val=val_data[:,3]
        w = val_data[:,4]
        

        m0, m1 = ICM(X_E=train_data[train_data[:,0]==1,1], X_O=train_data[train_data[:,0]==0,1], 
                     T_E=train_data[train_data[:,0]==1,2], T_O=train_data[train_data[:,0]==0,2], 
                     Y_E=train_data[train_data[:,0]==1,3], Y_O=train_data[train_data[:,0]==0,3],  
                     r=2, ID=1, AD=0, rho=rho)

        predY0_exp = m0.predict_noiseless(np.c_[np.vstack(val_data[:,1]), np.ones(val_data[:,1].shape[0]) * 0])
        predY1_exp = m1.predict_noiseless(np.c_[np.vstack(val_data[:,1]), np.ones(val_data[:,1].shape[0]) * 0])
        predCATE_exp = predY1_exp[0] - predY0_exp[0]
        varCATE_exp = predY1_exp[1] + predY0_exp[1]

        wmse = weighted_mean_squared_error(weight=w, t = T_val, y_true=Y_val, y_pred0=predY0_exp[0], y_pred1=predY1_exp[0])
        avg_wmse += wmse
        wmse_per_fold.append(wmse)

        print(f"Fold {fold_num + 1} WMSE: {wmse}")
        fold_num += 1
    
    avg_wmse /= num_folds
    print(f"Average WMSE for Rho {rho}: {avg_wmse}")
    wmse_per_rho[rho] = wmse_per_fold

    # Update min_avg_wmse and best_rho if current avg_wmse is lower
    if avg_wmse < min_avg_wmse:
        min_avg_wmse = avg_wmse
        best_rho_nobs250 = rho

# Print the best rho and its corresponding min_avg_wmse
print(f"Best Rho: {best_rho_nobs250}, Minimum Average WMSE: {min_avg_wmse}")

### Train model over 100 datasets using the optimal values of rho
rmse_nobs250 = np.zeros((100))
bias_nobs250 = np.zeros((100))
var_nobs250 = np.zeros((100))

for i in range(100):
    print(i + 1, end=" ")
    data_exp, data_obs = data_simulation(sample_size_exp=250, sample_size_obs=250)


    m0, m1 = ICM(X_E = data_exp[:,1], X_O = data_obs[:,1], 
                 T_E = data_exp[:,2], T_O = data_obs[:,2], 
                 Y_E = data_exp[:,3], Y_O = data_obs[:,3],  
                 r = 2, ID=1, AD=0, rho=best_rho_nobs250)
    predY0 = m0.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predY1 = m1.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predCATE = predY1[0] - predY0[0]

    

    rmse_nobs250[i] = np.sqrt(mean_squared_error((predCATE), (trueCATE)))
    bias_nobs250[i] = np.mean(predCATE - trueCATE)
    var_nobs250[i] = np.mean((predCATE - np.mean(predCATE))**2)



## $n_{exp} = 250, n_{obs} = 500$

### Tune $\rho$

from sklearn import linear_model
from sklearn.model_selection import KFold
from sklearn.metrics import mean_squared_error
import numpy.random as nprand
from neurIPS.functions.model_training import ICM
from neurIPS.functions.weightedMSE import weighted_mean_squared_error

data_exp, data_obs = data_simulation(sample_size_exp=250, sample_size_obs=500)
data_full = np.r_[data_exp, data_obs]

logr = linear_model.LogisticRegression()
logr.fit(data_full[:,1].reshape(-1,1),data_full[:,0])
prob = logr.predict_proba(data_full[:,1].reshape(-1,1))
prob_part = prob[:,1]
prob_nonpart = prob[:,0]
data_full = np.c_[data_full, 1/prob_part]

seed_value = 1234

# Define range of rho values to search over
rho_values = np.round(np.arange(0, 1.05, 0.1), 1)

# Define number of folds for cross-validation
num_folds = 5  # Example: 5-fold cross-validation

# Initialize variables to store results
cv_rmse_values = []
rho_rmse_values = {}  # Dictionary to store RMSE values for each rho
min_avg_wmse = float('inf')
best_rho_nobs500 = None
wmse_per_rho = {}

# Perform 5-fold cross-validation
kf = KFold(n_splits=num_folds, shuffle=True)
for rho in rho_values:
    
    print(f"Rho: {rho}")
    avg_wmse = 0
    fold_num = 0
    wmse_per_fold = []
    
    for train_index, val_index in kf.split(data_full):
        # 5 fold CV for experimental data: split them into 5 folds
        train_data, val_data= data_full[train_index], data_full[val_index]
        val_data = val_data[val_data[:,0]==1,:]
        X_val=val_data[:,1]
        T_val=val_data[:,2]
        Y_val=val_data[:,3]
        w = val_data[:,4]
        

        m0, m1 = ICM(X_E=train_data[train_data[:,0]==1,1], X_O=train_data[train_data[:,0]==0,1], 
                     T_E=train_data[train_data[:,0]==1,2], T_O=train_data[train_data[:,0]==0,2], 
                     Y_E=train_data[train_data[:,0]==1,3], Y_O=train_data[train_data[:,0]==0,3],  
                     r=2, ID=1, AD=0, rho=rho)

        predY0_exp = m0.predict_noiseless(np.c_[np.vstack(val_data[:,1]), np.ones(val_data[:,1].shape[0]) * 0])
        predY1_exp = m1.predict_noiseless(np.c_[np.vstack(val_data[:,1]), np.ones(val_data[:,1].shape[0]) * 0])
        predCATE_exp = predY1_exp[0] - predY0_exp[0]
        varCATE_exp = predY1_exp[1] + predY0_exp[1]

        wmse = weighted_mean_squared_error(weight=w, t = T_val, y_true=Y_val, y_pred0=predY0_exp[0], y_pred1=predY1_exp[0])
        avg_wmse += wmse
        wmse_per_fold.append(wmse)

        print(f"Fold {fold_num + 1} WMSE: {wmse}")
        fold_num += 1
    
    avg_wmse /= num_folds
    print(f"Average WMSE for Rho {rho}: {avg_wmse}")
    wmse_per_rho[rho] = wmse_per_fold

    # Update min_avg_wmse and best_rho if current avg_wmse is lower
    if avg_wmse < min_avg_wmse:
        min_avg_wmse = avg_wmse
        best_rho_nobs500 = rho

# Print the best rho and its corresponding min_avg_wmse
print(f"Best Rho: {best_rho_nobs500}, Minimum Average WMSE: {min_avg_wmse}")

### Train model over 100 datasets using the optimal value of $\rho$

rmse_nobs500 = np.zeros((100))
bias_nobs500 = np.zeros((100))
var_nobs500 = np.zeros((100))

for i in range(100):
    print(i + 1, end=" ")
    data_exp, data_obs = data_simulation(sample_size_exp=250, sample_size_obs=500)


    m0, m1 = ICM(X_E = data_exp[:,1], X_O = data_obs[:,1], 
                 T_E = data_exp[:,2], T_O = data_obs[:,2], 
                 Y_E = data_exp[:,3], Y_O = data_obs[:,3],  
                 r = 2, ID=1, AD=0, rho=best_rho_nobs500)
    predY0 = m0.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predY1 = m1.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predCATE = predY1[0] - predY0[0]

    

    rmse_nobs500[i] = np.sqrt(mean_squared_error((predCATE), (trueCATE)))
    bias_nobs500[i] = np.mean(predCATE - trueCATE)
    var_nobs500[i] = np.mean((predCATE - np.mean(predCATE))**2)


## $n_{exp}=250, n_{obs} = 1000$

### Tune $\rho$


data_exp, data_obs = data_simulation(sample_size_exp=250, sample_size_obs=1000)
data_full = np.r_[data_exp, data_obs]

logr = linear_model.LogisticRegression()
logr.fit(data_full[:,1].reshape(-1,1),data_full[:,0])
prob = logr.predict_proba(data_full[:,1].reshape(-1,1))
prob_part = prob[:,1]
prob_nonpart = prob[:,0]
data_full = np.c_[data_full, 1/prob_part]

seed_value = 1234

# Define range of rho values to search over
rho_values = np.round(np.arange(0, 1.05, 0.1), 1)

# Define number of folds for cross-validation
num_folds = 5  # Example: 5-fold cross-validation

# Initialize variables to store results
cv_rmse_values = []
rho_rmse_values = {}  # Dictionary to store RMSE values for each rho
min_avg_wmse = float('inf')
best_rho_nobs1000 = None
wmse_per_rho = {}

# Perform 5-fold cross-validation
kf = KFold(n_splits=num_folds, shuffle=True)
for rho in rho_values:
    
    print(f"Rho: {rho}")
    avg_wmse = 0
    fold_num = 0
    wmse_per_fold = []
    
    for train_index, val_index in kf.split(data_full):
        # 5 fold CV for experimental data: split them into 5 folds
        train_data, val_data= data_full[train_index], data_full[val_index]
        val_data = val_data[val_data[:,0]==1,:]
        X_val=val_data[:,1]
        T_val=val_data[:,2]
        Y_val=val_data[:,3]
        w = val_data[:,4]
        

        m0, m1 = ICM(X_E=train_data[train_data[:,0]==1,1], X_O=train_data[train_data[:,0]==0,1], 
                     T_E=train_data[train_data[:,0]==1,2], T_O=train_data[train_data[:,0]==0,2], 
                     Y_E=train_data[train_data[:,0]==1,3], Y_O=train_data[train_data[:,0]==0,3],  
                     r=2, ID=1, AD=0, rho=rho)

        predY0_exp = m0.predict_noiseless(np.c_[np.vstack(val_data[:,1]), np.ones(val_data[:,1].shape[0]) * 0])
        predY1_exp = m1.predict_noiseless(np.c_[np.vstack(val_data[:,1]), np.ones(val_data[:,1].shape[0]) * 0])
        predCATE_exp = predY1_exp[0] - predY0_exp[0]
        varCATE_exp = predY1_exp[1] + predY0_exp[1]

        wmse = weighted_mean_squared_error(weight=w, t = T_val, y_true=Y_val, y_pred0=predY0_exp[0], y_pred1=predY1_exp[0])
        avg_wmse += wmse
        wmse_per_fold.append(wmse)

        print(f"Fold {fold_num + 1} WMSE: {wmse}")
        fold_num += 1
    
    avg_wmse /= num_folds
    print(f"Average WMSE for Rho {rho}: {avg_wmse}")
    wmse_per_rho[rho] = wmse_per_fold

    # Update min_avg_wmse and best_rho if current avg_wmse is lower
    if avg_wmse < min_avg_wmse:
        min_avg_wmse = avg_wmse
        best_rho_nobs1000 = rho

# Print the best rho and its corresponding min_avg_wmse
print(f"Best Rho: {best_rho_nobs1000}, Minimum Average WMSE: {min_avg_wmse}")

### Train model over 100 datasets using the optimal value of $\rho$
rmse_nobs1000 = np.zeros((100))
bias_nobs1000 = np.zeros((100))
var_nobs1000 = np.zeros((100))

for i in range(100):
    print(i + 1, end=" ")
    data_exp, data_obs = data_simulation(sample_size_exp=250, sample_size_obs=1000)


    m0, m1 = ICM(X_E = data_exp[:,1], X_O = data_obs[:,1], 
                 T_E = data_exp[:,2], T_O = data_obs[:,2], 
                 Y_E = data_exp[:,3], Y_O = data_obs[:,3],  
                 r = 2, ID=1, AD=0, rho=best_rho_nobs1000)
    predY0 = m0.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predY1 = m1.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predCATE = predY1[0] - predY0[0]

    

    rmse_nobs1000[i] = np.sqrt(mean_squared_error((predCATE), (trueCATE)))
    bias_nobs1000[i] = np.mean(predCATE - trueCATE)
    var_nobs1000[i] = np.mean((predCATE - np.mean(predCATE))**2)


## $n_{exp}=250, n_{obs} = 2000$

### Tune $\rho$

from sklearn import linear_model
from sklearn.model_selection import KFold
from sklearn.metrics import mean_squared_error
import numpy.random as nprand
from neurIPS.functions.model_training import ICM
from neurIPS.functions.weightedMSE import weighted_mean_squared_error

data_exp, data_obs = data_simulation(sample_size_exp=250, sample_size_obs=2000)
data_full = np.r_[data_exp, data_obs]

logr = linear_model.LogisticRegression()
logr.fit(data_full[:,1].reshape(-1,1),data_full[:,0])
prob = logr.predict_proba(data_full[:,1].reshape(-1,1))
prob_part = prob[:,1]
prob_nonpart = prob[:,0]
data_full = np.c_[data_full, 1/prob_part]

seed_value = 1234

# Define range of rho values to search over
rho_values = np.round(np.arange(0, 1.05, 0.1), 1)

# Define number of folds for cross-validation
num_folds = 5  # Example: 5-fold cross-validation

# Initialize variables to store results
cv_rmse_values = []
rho_rmse_values = {}  # Dictionary to store RMSE values for each rho
min_avg_wmse = float('inf')
best_rho_nobs2000 = None
wmse_per_rho = {}

# Perform 5-fold cross-validation
kf = KFold(n_splits=num_folds, shuffle=True)
for rho in rho_values:
    
    print(f"Rho: {rho}")
    avg_wmse = 0
    fold_num = 0
    wmse_per_fold = []
    
    for train_index, val_index in kf.split(data_full):
        # 5 fold CV for experimental data: split them into 5 folds
        train_data, val_data= data_full[train_index], data_full[val_index]
        val_data = val_data[val_data[:,0]==1,:]
        X_val=val_data[:,1]
        T_val=val_data[:,2]
        Y_val=val_data[:,3]
        w = val_data[:,4]
        

        m0, m1 = ICM(X_E=train_data[train_data[:,0]==1,1], X_O=train_data[train_data[:,0]==0,1], 
                     T_E=train_data[train_data[:,0]==1,2], T_O=train_data[train_data[:,0]==0,2], 
                     Y_E=train_data[train_data[:,0]==1,3], Y_O=train_data[train_data[:,0]==0,3],  
                     r=2, ID=1, AD=0, rho=rho)

        predY0_exp = m0.predict_noiseless(np.c_[np.vstack(val_data[:,1]), np.ones(val_data[:,1].shape[0]) * 0])
        predY1_exp = m1.predict_noiseless(np.c_[np.vstack(val_data[:,1]), np.ones(val_data[:,1].shape[0]) * 0])
        predCATE_exp = predY1_exp[0] - predY0_exp[0]
        varCATE_exp = predY1_exp[1] + predY0_exp[1]

        wmse = weighted_mean_squared_error(weight=w, t = T_val, y_true=Y_val, y_pred0=predY0_exp[0], y_pred1=predY1_exp[0])
        avg_wmse += wmse
        wmse_per_fold.append(wmse)

        print(f"Fold {fold_num + 1} WMSE: {wmse}")
        fold_num += 1
    
    avg_wmse /= num_folds
    print(f"Average WMSE for Rho {rho}: {avg_wmse}")
    wmse_per_rho[rho] = wmse_per_fold

    # Update min_avg_wmse and best_rho if current avg_wmse is lower
    if avg_wmse < min_avg_wmse:
        min_avg_wmse = avg_wmse
        best_rho_nobs2000 = rho

# Print the best rho and its corresponding min_avg_wmse
print(f"Best Rho: {best_rho_nobs2000}, Minimum Average WMSE: {min_avg_wmse}")

### Train model over 100 datasets using the optimal value of $\rho$

rmse_nobs2000 = np.zeros((100))
bias_nobs2000 = np.zeros((100))
var_nobs2000 = np.zeros((100))

for i in range(100):
    print(i + 1, end=" ")
    data_exp, data_obs = data_simulation(sample_size_exp=250, sample_size_obs=2000)


    m0, m1 = ICM(X_E = data_exp[:,1], X_O = data_obs[:,1], 
                 T_E = data_exp[:,2], T_O = data_obs[:,2], 
                 Y_E = data_exp[:,3], Y_O = data_obs[:,3],  
                 r = 2, ID=1, AD=0, rho=best_rho_nobs2000)
    predY0 = m0.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predY1 = m1.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predCATE = predY1[0] - predY0[0]

    

    rmse_nobs2000[i] = np.sqrt(mean_squared_error((predCATE), (trueCATE)))
    bias_nobs2000[i] = np.mean(predCATE - trueCATE)
    var_nobs2000[i] = np.mean((predCATE - np.mean(predCATE))**2)


# Save the results
sens_nobs250 = pd.DataFrame({"RMSE": rmse_nobs250, "Bias": bias_nobs250, "Variance": var_nobs250})
sens_nobs500 = pd.DataFrame({"RMSE": rmse_nobs500, "Bias": bias_nobs500, "Variance": var_nobs500})
sens_nobs1000 = pd.DataFrame({"RMSE": rmse_nobs1000, "Bias": bias_nobs1000, "Variance": var_nobs1000})
sens_nobs2000 = pd.DataFrame({"RMSE": rmse_nobs2000, "Bias": bias_nobs2000, "Variance": var_nobs2000})


# Create a list of the DataFrames
dfs = [sens_nobs250, sens_nobs500, sens_nobs1000, sens_nobs2000]

# Create a list of rho values
nobs_values = [250, 500, 1000, 2000]

# Add the 'rho' column to each DataFrame
for df, nobs in zip(dfs, nobs_values):
    df['nobs'] = nobs

# Concatenate all the DataFrames into one
sens_nobs = pd.concat(dfs, ignore_index=True)

sens_nobs.to_csv('sens_nobs.csv', index=False)

# Display the combined DataFrame
print(sens_nobs)
print(best_rho_nobs250)
print(np.round(np.mean(rmse_nobs250), 3))
print(np.round(np.std(rmse_nobs250), 3))
print('\n')
print(best_rho_nobs500)
print(np.round(np.mean(rmse_nobs500), 3))
print(np.round(np.std(rmse_nobs500), 3))
print('\n')
print(best_rho_nobs1000)
print(np.round(np.mean(rmse_nobs1000), 3))
print(np.round(np.std(rmse_nobs1000), 3))
print('\n')
print(best_rho_nobs1000)
print(np.round(np.mean(rmse_nobs2000), 3))
print(np.round(np.std(rmse_nobs2000), 3))


# Degree of overlap

## Full overlap

def data_simulation_full_ovelap(sample_size_exp, sample_size_obs):
    
    # RCT
    X_trial = np.random.uniform(-2, 2.0, sample_size_exp)
    S_trial = np.ones(sample_size_exp)
    A_trial = bernoulli.rvs(0.5, size=sample_size_exp)
    tau_X_trial = 1 + X_trial + X_trial**2
    epsilon_trial = norm.rvs(0, 1.0, sample_size_exp)
    Y_trial = A_trial * tau_X_trial + X_trial**2 - 1 + epsilon_trial

    # Observational study
    X_observational = np.random.uniform(-2, 2, sample_size_obs)
    S_observational = np.zeros(sample_size_obs)
    logit_e_X_0 = X_observational
    A_observational = bernoulli.rvs(1 / (1 + np.exp(-logit_e_X_0)), size=sample_size_obs)
    tau_X_observational = 1 + X_observational + X_observational**2
    U_observational = (2 * A_observational - 1) *np.sin(X_observational-1) + norm.rvs(0, 1, size=sample_size_obs)
    epsilon_observational = norm.rvs(0, 1, size=sample_size_obs)
    Y_observational = A_observational * tau_X_observational + X_observational**2 - 1 + U_observational + epsilon_observational

    # Create DataFrames
    data_trial = np.column_stack((S_trial, X_trial, A_trial, Y_trial))
    data_observational = np.column_stack((S_observational, X_observational, A_observational, Y_observational))

    return data_trial, data_observational



data_exp, data_obs = data_simulation_full_ovelap(sample_size_exp=250, sample_size_obs=1000)
data_full = np.r_[data_exp, data_obs]

logr = linear_model.LogisticRegression()
logr.fit(data_full[:,1].reshape(-1,1),data_full[:,0])
prob = logr.predict_proba(data_full[:,1].reshape(-1,1))
prob_part = prob[:,1]
prob_nonpart = prob[:,0]
data_full = np.c_[data_full, 1/prob_part]

seed_value = 1234

# Define range of rho values to search over
rho_values = np.round(np.arange(0, 1.05, 0.1), 1)

# Define number of folds for cross-validation
num_folds = 5  # Example: 5-fold cross-validation

# Initialize variables to store results
cv_rmse_values = []
rho_rmse_values = {}  # Dictionary to store RMSE values for each rho
min_avg_wmse = float('inf')
best_rho_full_overlap = None
wmse_per_rho = {}

# Perform 5-fold cross-validation
kf = KFold(n_splits=num_folds, shuffle=True)
for rho in rho_values:
    
    print(f"Rho: {rho}")
    avg_wmse = 0
    fold_num = 0
    wmse_per_fold = []
    
    for train_index, val_index in kf.split(data_full):
        # 5 fold CV for experimental data: split them into 5 folds
        train_data, val_data= data_full[train_index], data_full[val_index]
        val_data = val_data[val_data[:,0]==1,:]
        X_val=val_data[:,1]
        T_val=val_data[:,2]
        Y_val=val_data[:,3]
        w = val_data[:,4]
        

        m0, m1 = ICM(X_E=train_data[train_data[:,0]==1,1], X_O=train_data[train_data[:,0]==0,1], 
                     T_E=train_data[train_data[:,0]==1,2], T_O=train_data[train_data[:,0]==0,2], 
                     Y_E=train_data[train_data[:,0]==1,3], Y_O=train_data[train_data[:,0]==0,3],  
                     r=2, ID=1, AD=0, rho=rho)

        predY0_exp = m0.predict_noiseless(np.c_[np.vstack(val_data[:,1]), np.ones(val_data[:,1].shape[0]) * 0])
        predY1_exp = m1.predict_noiseless(np.c_[np.vstack(val_data[:,1]), np.ones(val_data[:,1].shape[0]) * 0])
        predCATE_exp = predY1_exp[0] - predY0_exp[0]
        varCATE_exp = predY1_exp[1] + predY0_exp[1]

        wmse = weighted_mean_squared_error(weight=w, t = T_val, y_true=Y_val, y_pred0=predY0_exp[0], y_pred1=predY1_exp[0])
        avg_wmse += wmse
        wmse_per_fold.append(wmse)

        print(f"Fold {fold_num + 1} WMSE: {wmse}")
        fold_num += 1
    
    avg_wmse /= num_folds
    print(f"Average WMSE for Rho {rho}: {avg_wmse}")
    wmse_per_rho[rho] = wmse_per_fold

    # Update min_avg_wmse and best_rho if current avg_wmse is lower
    if avg_wmse < min_avg_wmse:
        min_avg_wmse = avg_wmse
        best_rho_full_overlap = rho

# Print the best rho and its corresponding min_avg_wmse
print(f"Best Rho: {best_rho_full_overlap}, Minimum Average WMSE: {min_avg_wmse}")
rmse_full_overlap = np.zeros((100))
bias_full_overlap = np.zeros((100))
var_full_overlap = np.zeros((100))

for i in range(100):
    print(i + 1, end=" ")
    data_exp, data_obs = data_simulation_full_ovelap(sample_size_exp=250, sample_size_obs=1000)


    m0, m1 = ICM(X_E = data_exp[:,1], X_O = data_obs[:,1], 
                 T_E = data_exp[:,2], T_O = data_obs[:,2], 
                 Y_E = data_exp[:,3], Y_O = data_obs[:,3],  
                 r = 2, ID=1, AD=0, rho=best_rho_full_overlap)
    predY0 = m0.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predY1 = m1.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predCATE = predY1[0] - predY0[0]

    

    rmse_full_overlap[i] = np.sqrt(mean_squared_error((predCATE), (trueCATE)))
    bias_full_overlap[i] = np.mean(predCATE - trueCATE)
    var_full_overlap[i] = np.mean((predCATE - np.mean(predCATE))**2)


## High overlap (but not full)

def data_simulation_high_ovelap(sample_size_exp, sample_size_obs):
    
    # RCT
    X_trial = np.random.uniform(-2, 1.0, sample_size_exp)
    S_trial = np.ones(sample_size_exp)
    A_trial = bernoulli.rvs(0.5, size=sample_size_exp)
    tau_X_trial = 1 + X_trial + X_trial**2
    epsilon_trial = norm.rvs(0, 1.0, sample_size_exp)
    Y_trial = A_trial * tau_X_trial + X_trial**2 - 1 + epsilon_trial

    # Observational study
    X_observational = np.random.uniform(-2, 2, sample_size_obs)
    S_observational = np.zeros(sample_size_obs)
    logit_e_X_0 = X_observational
    A_observational = bernoulli.rvs(1 / (1 + np.exp(-logit_e_X_0)), size=sample_size_obs)
    tau_X_observational = 1 + X_observational + X_observational**2
    U_observational = (2 * A_observational - 1) *np.sin(X_observational-1) + norm.rvs(0, 1, size=sample_size_obs)
    epsilon_observational = norm.rvs(0, 1, size=sample_size_obs)
    Y_observational = A_observational * tau_X_observational + X_observational**2 - 1 + U_observational + epsilon_observational

    # Create DataFrames
    data_trial = np.column_stack((S_trial, X_trial, A_trial, Y_trial))
    data_observational = np.column_stack((S_observational, X_observational, A_observational, Y_observational))

    return data_trial, data_observational


data_exp, data_obs = data_simulation_high_ovelap(sample_size_exp=250, sample_size_obs=1000)
data_full = np.r_[data_exp, data_obs]

logr = linear_model.LogisticRegression()
logr.fit(data_full[:,1].reshape(-1,1),data_full[:,0])
prob = logr.predict_proba(data_full[:,1].reshape(-1,1))
prob_part = prob[:,1]
prob_nonpart = prob[:,0]
data_full = np.c_[data_full, 1/prob_part]

seed_value = 1234

# Define range of rho values to search over
rho_values = np.round(np.arange(0, 1.05, 0.1), 1)

# Define number of folds for cross-validation
num_folds = 5  # Example: 5-fold cross-validation

# Initialize variables to store results
cv_rmse_values = []
rho_rmse_values = {}  # Dictionary to store RMSE values for each rho
min_avg_wmse = float('inf')
best_rho_high_overlap = None
wmse_per_rho = {}

# Perform 5-fold cross-validation
kf = KFold(n_splits=num_folds, shuffle=True)
for rho in rho_values:
    
    print(f"Rho: {rho}")
    avg_wmse = 0
    fold_num = 0
    wmse_per_fold = []
    
    for train_index, val_index in kf.split(data_full):
        # 5 fold CV for experimental data: split them into 5 folds
        train_data, val_data= data_full[train_index], data_full[val_index]
        val_data = val_data[val_data[:,0]==1,:]
        X_val=val_data[:,1]
        T_val=val_data[:,2]
        Y_val=val_data[:,3]
        w = val_data[:,4]
        

        m0, m1 = ICM(X_E=train_data[train_data[:,0]==1,1], X_O=train_data[train_data[:,0]==0,1], 
                     T_E=train_data[train_data[:,0]==1,2], T_O=train_data[train_data[:,0]==0,2], 
                     Y_E=train_data[train_data[:,0]==1,3], Y_O=train_data[train_data[:,0]==0,3],  
                     r=2, ID=1, AD=0, rho=rho)

        predY0_exp = m0.predict_noiseless(np.c_[np.vstack(val_data[:,1]), np.ones(val_data[:,1].shape[0]) * 0])
        predY1_exp = m1.predict_noiseless(np.c_[np.vstack(val_data[:,1]), np.ones(val_data[:,1].shape[0]) * 0])
        predCATE_exp = predY1_exp[0] - predY0_exp[0]
        varCATE_exp = predY1_exp[1] + predY0_exp[1]

        wmse = weighted_mean_squared_error(weight=w, t = T_val, y_true=Y_val, y_pred0=predY0_exp[0], y_pred1=predY1_exp[0])
        avg_wmse += wmse
        wmse_per_fold.append(wmse)

        print(f"Fold {fold_num + 1} WMSE: {wmse}")
        fold_num += 1
    
    avg_wmse /= num_folds
    print(f"Average WMSE for Rho {rho}: {avg_wmse}")
    wmse_per_rho[rho] = wmse_per_fold

    # Update min_avg_wmse and best_rho if current avg_wmse is lower
    if avg_wmse < min_avg_wmse:
        min_avg_wmse = avg_wmse
        best_rho_high_overlap = rho


# Print the best rho and its corresponding min_avg_wmse
print(f"Best Rho: {best_rho_high_overlap}, Minimum Average WMSE: {min_avg_wmse}")
rmse_high_overlap = np.zeros((100))
bias_high_overlap = np.zeros((100))
var_high_overlap = np.zeros((100))

for i in range(100):
    print(i + 1, end=" ")
    data_exp, data_obs = data_simulation_high_ovelap(sample_size_exp=250, sample_size_obs=1000)


    m0, m1 = ICM(X_E = data_exp[:,1], X_O = data_obs[:,1], 
                 T_E = data_exp[:,2], T_O = data_obs[:,2], 
                 Y_E = data_exp[:,3], Y_O = data_obs[:,3],  
                 r = 2, ID=1, AD=0, rho=best_rho_high_overlap)
    predY0 = m0.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predY1 = m1.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predCATE = predY1[0] - predY0[0]

    

    rmse_high_overlap[i] = np.sqrt(mean_squared_error((predCATE), (trueCATE)))
    bias_high_overlap[i] = np.mean(predCATE - trueCATE)
    var_high_overlap[i] = np.mean((predCATE - np.mean(predCATE))**2)


## Low overlap

def data_simulation_low_ovelap(sample_size_exp, sample_size_obs):
    
    # RCT
    X_trial = np.random.uniform(-2, -0.5, sample_size_exp)
    S_trial = np.ones(sample_size_exp)
    A_trial = bernoulli.rvs(0.5, size=sample_size_exp)
    tau_X_trial = 1 + X_trial + X_trial**2
    epsilon_trial = norm.rvs(0, 1.0, sample_size_exp)
    Y_trial = A_trial * tau_X_trial + X_trial**2 - 1 + epsilon_trial

    # Observational study
    X_observational = np.random.uniform(-2, 2, sample_size_obs)
    S_observational = np.zeros(sample_size_obs)
    logit_e_X_0 = X_observational
    A_observational = bernoulli.rvs(1 / (1 + np.exp(-logit_e_X_0)), size=sample_size_obs)
    tau_X_observational = 1 + X_observational + X_observational**2
    U_observational = (2 * A_observational - 1) *np.sin(X_observational-1) + norm.rvs(0, 1, size=sample_size_obs)
    epsilon_observational = norm.rvs(0, 1, size=sample_size_obs)
    Y_observational = A_observational * tau_X_observational + X_observational**2 - 1 + U_observational + epsilon_observational

    # Create DataFrames
    data_trial = np.column_stack((S_trial, X_trial, A_trial, Y_trial))
    data_observational = np.column_stack((S_observational, X_observational, A_observational, Y_observational))

    return data_trial, data_observational


data_exp, data_obs = data_simulation_low_ovelap(sample_size_exp=250, sample_size_obs=1000)
data_full = np.r_[data_exp, data_obs]

logr = linear_model.LogisticRegression()
logr.fit(data_full[:,1].reshape(-1,1),data_full[:,0])
prob = logr.predict_proba(data_full[:,1].reshape(-1,1))
prob_part = prob[:,1]
prob_nonpart = prob[:,0]
data_full = np.c_[data_full, 1/prob_part]

seed_value = 1234

# Define range of rho values to search over
rho_values = np.round(np.arange(0, 1.05, 0.1), 1)

# Define number of folds for cross-validation
num_folds = 5  # Example: 5-fold cross-validation

# Initialize variables to store results
cv_rmse_values = []
rho_rmse_values = {}  # Dictionary to store RMSE values for each rho
min_avg_wmse = float('inf')
best_rho_low_overlap = None
wmse_per_rho = {}

# Perform 5-fold cross-validation
kf = KFold(n_splits=num_folds, shuffle=True)
for rho in rho_values:
    
    print(f"Rho: {rho}")
    avg_wmse = 0
    fold_num = 0
    wmse_per_fold = []
    
    for train_index, val_index in kf.split(data_full):
        # 5 fold CV for experimental data: split them into 5 folds
        train_data, val_data= data_full[train_index], data_full[val_index]
        val_data = val_data[val_data[:,0]==1,:]
        X_val=val_data[:,1]
        T_val=val_data[:,2]
        Y_val=val_data[:,3]
        w = val_data[:,4]
        

        m0, m1 = ICM(X_E=train_data[train_data[:,0]==1,1], X_O=train_data[train_data[:,0]==0,1], 
                     T_E=train_data[train_data[:,0]==1,2], T_O=train_data[train_data[:,0]==0,2], 
                     Y_E=train_data[train_data[:,0]==1,3], Y_O=train_data[train_data[:,0]==0,3],  
                     r=2, ID=1, AD=0, rho=rho)

        predY0_exp = m0.predict_noiseless(np.c_[np.vstack(val_data[:,1]), np.ones(val_data[:,1].shape[0]) * 0])
        predY1_exp = m1.predict_noiseless(np.c_[np.vstack(val_data[:,1]), np.ones(val_data[:,1].shape[0]) * 0])
        predCATE_exp = predY1_exp[0] - predY0_exp[0]
        varCATE_exp = predY1_exp[1] + predY0_exp[1]

        wmse = weighted_mean_squared_error(weight=w, t = T_val, y_true=Y_val, y_pred0=predY0_exp[0], y_pred1=predY1_exp[0])
        avg_wmse += wmse
        wmse_per_fold.append(wmse)

        print(f"Fold {fold_num + 1} WMSE: {wmse}")
        fold_num += 1
    
    avg_wmse /= num_folds
    print(f"Average WMSE for Rho {rho}: {avg_wmse}")
    wmse_per_rho[rho] = wmse_per_fold

    # Update min_avg_wmse and best_rho if current avg_wmse is lower
    if avg_wmse < min_avg_wmse:
        min_avg_wmse = avg_wmse
        best_rho_low_overlap = rho


# Print the best rho and its corresponding min_avg_wmse

print(f"Best Rho: {best_rho_low_overlap}, Minimum Average WMSE: {min_avg_wmse}")
rmse_low_overlap = np.zeros((100))
bias_low_overlap = np.zeros((100))
var_low_overlap = np.zeros((100))

for i in range(100):
    print(i + 1, end=" ")
    data_exp, data_obs = data_simulation_low_ovelap(sample_size_exp=250, sample_size_obs=1000)


    m0, m1 = ICM(X_E = data_exp[:,1], X_O = data_obs[:,1], 
                 T_E = data_exp[:,2], T_O = data_obs[:,2], 
                 Y_E = data_exp[:,3], Y_O = data_obs[:,3],  
                 r = 2, ID=1, AD=0, rho=best_rho_low_overlap)
    predY0 = m0.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predY1 = m1.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predCATE = predY1[0] - predY0[0]

    

    rmse_low_overlap[i] = np.sqrt(mean_squared_error((predCATE), (trueCATE)))
    bias_low_overlap[i] = np.mean(predCATE - trueCATE)
    var_low_overlap[i] = np.mean((predCATE - np.mean(predCATE))**2)


print(best_rho_full_overlap)
print(np.round(np.mean(rmse_full_overlap), 3))
print(np.round(np.std(rmse_full_overlap), 3))
print('\n')

print(best_rho_high_overlap)
print(np.round(np.mean(rmse_high_overlap), 3))
print(np.round(np.std(rmse_high_overlap), 3))
print('\n')

print(best_rho_low_overlap)
print(np.round(np.mean(rmse_low_overlap), 3))
print(np.round(np.std(rmse_low_overlap), 3))
print('\n')

# Save the results
sens_ful_overlap = pd.DataFrame({"RMSE": rmse_full_overlap, "Bias": bias_full_overlap, "Variance": var_full_overlap})
sens_high_overlap = pd.DataFrame({"RMSE": rmse_high_overlap, "Bias": bias_high_overlap, "Variance": var_high_overlap})
sens_low_overlap = pd.DataFrame({"RMSE": rmse_low_overlap, "Bias": bias_low_overlap, "Variance": var_low_overlap})

# Create a list of the DataFrames
dfs = [sens_ful_overlap, sens_high_overlap, sens_low_overlap]

# Create a list of overlap values
overlap_values = ['full', 'high', 'low']

# Add the 'overlap' column to each DataFrame
for df, overlap in zip(dfs, overlap_values):
    df['overlap'] = overlap

# Concatenate all the DataFrames into one
sens_overlap = pd.concat(dfs, ignore_index=True)

# Save the combined DataFrame to a CSV file
sens_overlap.to_csv('sens_overlap.csv', index=False)

# Display the combined DataFrame
print(sens_overlap)

#################################################################################

# Kernel Choice

## RBF kernel

rho_RBF = pd.read_csv('/path/to/your/CausalICM/Simulation 8/sim8_best_rho.csv').values[0]

rmse_RBF = np.zeros((100))
bias_RBF = np.zeros((100))
var_RBF = np.zeros((100))


for i in range(100):
    print(i + 1, end=" ")
    data_exp = datasets[(datasets['S'] == 1) & (datasets['datasetNo'] == i + 1)]
    data_obs = datasets[(datasets['S'] == 0) & (datasets['datasetNo'] == i + 1)]


    m0, m1 = ICM(X_E = data_exp["X"], X_O = data_obs["X"], 
                 T_E = data_exp["A"], T_O = data_obs["A"], 
                 Y_E = data_exp["Y"], Y_O = data_obs["Y"],  
                 r = 2, ID=1, AD=0, rho=rho_RBF)
    predY0 = m0.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predY1 = m1.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predCATE = predY1[0] - predY0[0]

    

    rmse_RBF[i] = np.sqrt(mean_squared_error((predCATE), (trueCATE)))
    bias_RBF[i] = np.mean(predCATE - trueCATE)
    var_RBF[i] = np.mean((predCATE - np.mean(predCATE))**2)

## Matérn 3/2 kernel


def ICM_matern32(X_E, X_O, T_E, T_O, Y_E, Y_O, r, ID, AD, rho):
    """
    Train and return two rank-2 ICMs
    Args:
    - X_E (numpy.ndarray): Baseline covariate in the experimental data.
    - X_O (numpy.ndarray): Baseline covariate in the observational data.
    - T_E (numpy.ndarray): Treatment indicator in the experimental data.
    - T_O (numpy.ndarray): Treatment indicator in the observational data.
    - Y_E (numpy.ndarray): Outcome in the experimental data.
    - Y_O (numpy.ndarray): Outcome in the observational data.
    - r (int): Rank of the model.
    - ID (int): input dim
    - AD (list): active dim
    - rho: trust hyperparameter

    Returns:
    GPy.models.GPRegression: Trained Gaussian Process Regression model using Intrinsic Coregionalized Model (ICM).

    Data Processing:
    - Split the variables into treated and control for both observational and experimental data.

    Model Training:
    - Intrinsic Coregionalized Model (ICM) with a Radial Basis Function (RBF) kernel is used for regression.
    - The model is trained using Gaussian Process Regression.

    Note:
    - The order of arranging data matters: Experimental control, Observational control, Experimental treated, Observational treated.
    - An indicator is added to the baseline covariate to represent the task (0 for experimental control, 1 for observational control, 2 for experimental treated, 3 for observational treated).
    """

    # Data preprocessing
    # Split the variables in treated and control 
    # Observational Data
    X0_obs = X_O[T_O==0] # baseline covariate in th control arm of the observational study
    X1_obs = X_O[T_O==1] # baseline covariate in th treatment arm of the observational study
    Y0_obs = Y_O[T_O==0] # outcomee in th control arm of the observational study
    Y1_obs = Y_O[T_O==1] # outcomee in th treatment arm of the observational study
    
    #Exoerimental data
    X0_exp = X_E[T_E==0] # baseline covariate in th control arm of the experimental study
    X1_exp = X_E[T_E==1] # baseline covariate in th treatment arm of the experimental study
    Y0_exp = Y_E[T_E==0] # outcomee in th control arm of the experimental study
    Y1_exp = Y_E[T_E==1] # outcomee in th treatment arm of the experimental study
    

    # The order we arrange the data matters
    # 1. Experimental control
    # 2. Observational control
    # 3. Experimental treated
    # 4. Observational treated
    # We also need to add an indicator to the baseline covariate indicating the task
    # i.e. 0 is for the experiemental control, 1 is for the observational control,
    # 2 is for the experiemental treated, 3 is for observational treated
    X0_exp = np.c_[X0_exp,np.ones(X0_exp.shape[0])*0] 
    X0_obs = np.c_[X0_obs,np.ones(X0_obs.shape[0])*1]
    X1_exp = np.c_[X1_exp,np.ones(X1_exp.shape[0])*0]
    X1_obs = np.c_[X1_obs,np.ones(X1_obs.shape[0])*1]

    # Put all the data on a single array 
    Y0 = np.r_[Y0_exp, Y0_obs]
    X0 = np.r_[X0_exp, X0_obs]
    Y1 = np.r_[Y1_exp, Y1_obs]
    X1 = np.r_[X1_exp, X1_obs]


    # model training
    
    # Define the kernel
    kern = GPy.kern.Matern32(input_dim=ID, active_dims = AD)**GPy.kern.Coregionalize(input_dim = 1, output_dim = 2, rank=r)
    # Define the model
    m0 = GPy.models.GPRegression(X0,np.vstack(Y0),kern)
    ##############################################################
    # Put all your constraines here
    m0.mul.coregion.W[0,0:1].fix(1)
    m0.mul.coregion.W[0,1:].fix(0)
    m0.mul.coregion.W[1,0:1].fix(rho)
    m0.mul.coregion.W[1,1:].fix(np.sqrt(1-rho**2))
    m0.mul.coregion.kappa.fix([0.000000001, 0.000000001])
    #m0.mul.rbf.variance.constrain_bounded(0., 10.)
    
    ##############################################################
    

    # Optimize the model
    m0.optimize()

    # Define the kernel
    kern = GPy.kern.Matern32(input_dim=ID, active_dims = AD)**GPy.kern.Coregionalize(input_dim = 1, output_dim = 2, rank=r)
    # Define the model
    m1 = GPy.models.GPRegression(X1,np.vstack(Y1),kern)
    ##############################################################
    # Put all your constraines here
    m1.mul.coregion.W[0,0:1].fix(1)
    m1.mul.coregion.W[0,1:].fix(0)
    m1.mul.coregion.W[1,0:1].fix(rho)
    m1.mul.coregion.W[1,1:].fix(np.sqrt(1-rho**2))
    m1.mul.coregion.kappa.fix([0.000000001, 0.000000001])
    #m1.mul.rbf.variance.constrain_bounded(0., 10.)

    ##############################################################
    # Optimize the model
    
    m1.optimize()

    return m0, m1



data_exp, data_obs = data_simulation8(1000, beta0=-3.0, beta1=-3.0)
data_full = np.r_[data_exp, data_obs]

logr = linear_model.LogisticRegression()
logr.fit(data_full[:,1].reshape(-1,1),data_full[:,0])
prob = logr.predict_proba(data_full[:,1].reshape(-1,1))
prob_part = prob[:,1]
prob_nonpart = prob[:,0]
data_full = np.c_[data_full, 1/prob_part]

seed_value = 1234

# Define range of rho values to search over
rho_values = np.round(np.arange(0, 1.05, 0.1), 1)

# Define number of folds for cross-validation
num_folds = 5  # Example: 5-fold cross-validation

# Initialize variables to store results
cv_rmse_values = []
rho_rmse_values = {}  # Dictionary to store RMSE values for each rho
min_avg_wmse = float('inf')
best_rho_matern32 = None
wmse_per_rho = {}

# Perform 5-fold cross-validation
kf = KFold(n_splits=num_folds, shuffle=True)
for rho in rho_values:
    
    print(f"Rho: {rho}")
    avg_wmse = 0
    fold_num = 0
    wmse_per_fold = []
    
    for train_index, val_index in kf.split(data_full):
        # 5 fold CV for experimental data: split them into 5 folds
        train_data, val_data= data_full[train_index], data_full[val_index]
        val_data = val_data[val_data[:,0]==1,:]
        X_val=val_data[:,1]
        T_val=val_data[:,2]
        Y_val=val_data[:,3]
        w = val_data[:,4]
        

        m0, m1 = ICM_matern32(X_E=train_data[train_data[:,0]==1,1], X_O=train_data[train_data[:,0]==0,1], 
                     T_E=train_data[train_data[:,0]==1,2], T_O=train_data[train_data[:,0]==0,2], 
                     Y_E=train_data[train_data[:,0]==1,3], Y_O=train_data[train_data[:,0]==0,3],  
                     r=2, ID=1, AD=0, rho=rho)

        predY0_exp = m0.predict_noiseless(np.c_[np.vstack(val_data[:,1]), np.ones(val_data[:,1].shape[0]) * 0])
        predY1_exp = m1.predict_noiseless(np.c_[np.vstack(val_data[:,1]), np.ones(val_data[:,1].shape[0]) * 0])
        predCATE_exp = predY1_exp[0] - predY0_exp[0]
        varCATE_exp = predY1_exp[1] + predY0_exp[1]

        wmse = weighted_mean_squared_error(weight=w, t = T_val, y_true=Y_val, y_pred0=predY0_exp[0], y_pred1=predY1_exp[0])
        avg_wmse += wmse
        wmse_per_fold.append(wmse)

        print(f"Fold {fold_num + 1} WMSE: {wmse}")
        fold_num += 1
    
    avg_wmse /= num_folds
    print(f"Average WMSE for Rho {rho}: {avg_wmse}")
    wmse_per_rho[rho] = wmse_per_fold

    # Update min_avg_wmse and best_rho if current avg_wmse is lower
    if avg_wmse < min_avg_wmse:
        min_avg_wmse = avg_wmse
        best_rho_matern32 = rho

# Print the best rho and its corresponding min_avg_wmse
print(f"Best Rho: {best_rho_matern32}, Minimum Average WMSE: {min_avg_wmse}")
rmse_matern32 = np.zeros((100))
bias_matern32 = np.zeros((100))
var_matern32 = np.zeros((100))


for i in range(100):
    print(i + 1, end=" ")
    data_exp = datasets[(datasets['S'] == 1) & (datasets['datasetNo'] == i + 1)]
    data_obs = datasets[(datasets['S'] == 0) & (datasets['datasetNo'] == i + 1)]


    m0, m1 = ICM_matern32(X_E = data_exp["X"], X_O = data_obs["X"], 
                 T_E = data_exp["A"], T_O = data_obs["A"], 
                 Y_E = data_exp["Y"], Y_O = data_obs["Y"],  
                 r = 2, ID=1, AD=0, rho=best_rho_matern32)
    predY0 = m0.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predY1 = m1.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predCATE = predY1[0] - predY0[0]

    

    rmse_matern32[i] = np.sqrt(mean_squared_error((predCATE), (trueCATE)))
    bias_matern32[i] = np.mean(predCATE - trueCATE)
    var_matern32[i] = np.mean((predCATE - np.mean(predCATE))**2)


## Matérn 5/2 kernel

def ICM_matern52(X_E, X_O, T_E, T_O, Y_E, Y_O, r, ID, AD, rho):
    """
    Train and return two rank-2 ICMs
    Args:
    - X_E (numpy.ndarray): Baseline covariate in the experimental data.
    - X_O (numpy.ndarray): Baseline covariate in the observational data.
    - T_E (numpy.ndarray): Treatment indicator in the experimental data.
    - T_O (numpy.ndarray): Treatment indicator in the observational data.
    - Y_E (numpy.ndarray): Outcome in the experimental data.
    - Y_O (numpy.ndarray): Outcome in the observational data.
    - r (int): Rank of the model.
    - ID (int): input dim
    - AD (list): active dim
    - rho: trust hyperparameter

    Returns:
    GPy.models.GPRegression: Trained Gaussian Process Regression model using Intrinsic Coregionalized Model (ICM).

    Data Processing:
    - Split the variables into treated and control for both observational and experimental data.

    Model Training:
    - Intrinsic Coregionalized Model (ICM) with a Radial Basis Function (RBF) kernel is used for regression.
    - The model is trained using Gaussian Process Regression.

    Note:
    - The order of arranging data matters: Experimental control, Observational control, Experimental treated, Observational treated.
    - An indicator is added to the baseline covariate to represent the task (0 for experimental control, 1 for observational control, 2 for experimental treated, 3 for observational treated).
    """

    # Data preprocessing
    # Split the variables in treated and control 
    # Observational Data
    X0_obs = X_O[T_O==0] # baseline covariate in th control arm of the observational study
    X1_obs = X_O[T_O==1] # baseline covariate in th treatment arm of the observational study
    Y0_obs = Y_O[T_O==0] # outcomee in th control arm of the observational study
    Y1_obs = Y_O[T_O==1] # outcomee in th treatment arm of the observational study
    
    #Exoerimental data
    X0_exp = X_E[T_E==0] # baseline covariate in th control arm of the experimental study
    X1_exp = X_E[T_E==1] # baseline covariate in th treatment arm of the experimental study
    Y0_exp = Y_E[T_E==0] # outcomee in th control arm of the experimental study
    Y1_exp = Y_E[T_E==1] # outcomee in th treatment arm of the experimental study
    

    # The order we arrange the data matters
    # 1. Experimental control
    # 2. Observational control
    # 3. Experimental treated
    # 4. Observational treated
    # We also need to add an indicator to the baseline covariate indicating the task
    # i.e. 0 is for the experiemental control, 1 is for the observational control,
    # 2 is for the experiemental treated, 3 is for observational treated
    X0_exp = np.c_[X0_exp,np.ones(X0_exp.shape[0])*0] 
    X0_obs = np.c_[X0_obs,np.ones(X0_obs.shape[0])*1]
    X1_exp = np.c_[X1_exp,np.ones(X1_exp.shape[0])*0]
    X1_obs = np.c_[X1_obs,np.ones(X1_obs.shape[0])*1]

    # Put all the data on a single array 
    Y0 = np.r_[Y0_exp, Y0_obs]
    X0 = np.r_[X0_exp, X0_obs]
    Y1 = np.r_[Y1_exp, Y1_obs]
    X1 = np.r_[X1_exp, X1_obs]


    # model training
    
    # Define the kernel
    kern = GPy.kern.Matern52(input_dim=ID, active_dims = AD)**GPy.kern.Coregionalize(input_dim = 1, output_dim = 2, rank=r)
    # Define the model
    m0 = GPy.models.GPRegression(X0,np.vstack(Y0),kern)
    ##############################################################
    # Put all your constraines here
    m0.mul.coregion.W[0,0:1].fix(1)
    m0.mul.coregion.W[0,1:].fix(0)
    m0.mul.coregion.W[1,0:1].fix(rho)
    m0.mul.coregion.W[1,1:].fix(np.sqrt(1-rho**2))
    m0.mul.coregion.kappa.fix([0.000000001, 0.000000001])
    #m0.mul.rbf.variance.constrain_bounded(0., 10.)
    
    ##############################################################
    

    # Optimize the model
    m0.optimize()

    # Define the kernel
    kern = GPy.kern.Matern52(input_dim=ID, active_dims = AD)**GPy.kern.Coregionalize(input_dim = 1, output_dim = 2, rank=r)
    # Define the model
    m1 = GPy.models.GPRegression(X1,np.vstack(Y1),kern)
    ##############################################################
    # Put all your constraines here
    m1.mul.coregion.W[0,0:1].fix(1)
    m1.mul.coregion.W[0,1:].fix(0)
    m1.mul.coregion.W[1,0:1].fix(rho)
    m1.mul.coregion.W[1,1:].fix(np.sqrt(1-rho**2))
    m1.mul.coregion.kappa.fix([0.000000001, 0.000000001])
    #m1.mul.rbf.variance.constrain_bounded(0., 10.)

    ##############################################################
    # Optimize the model
    
    m1.optimize()

    return m0, m1



data_exp, data_obs = data_simulation8(1000, beta0=-3.0, beta1=-3.0)
data_full = np.r_[data_exp, data_obs]

logr = linear_model.LogisticRegression()
logr.fit(data_full[:,1].reshape(-1,1),data_full[:,0])
prob = logr.predict_proba(data_full[:,1].reshape(-1,1))
prob_part = prob[:,1]
prob_nonpart = prob[:,0]
data_full = np.c_[data_full, 1/prob_part]

seed_value = 1234

# Define range of rho values to search over
rho_values = np.round(np.arange(0, 1.05, 0.1), 1)

# Define number of folds for cross-validation
num_folds = 5  # Example: 5-fold cross-validation

# Initialize variables to store results
cv_rmse_values = []
rho_rmse_values = {}  # Dictionary to store RMSE values for each rho
min_avg_wmse = float('inf')
best_rho_matern52 = None
wmse_per_rho = {}

# Perform 5-fold cross-validation
kf = KFold(n_splits=num_folds, shuffle=True)
for rho in rho_values:
    
    print(f"Rho: {rho}")
    avg_wmse = 0
    fold_num = 0
    wmse_per_fold = []
    
    for train_index, val_index in kf.split(data_full):
        # 5 fold CV for experimental data: split them into 5 folds
        train_data, val_data= data_full[train_index], data_full[val_index]
        val_data = val_data[val_data[:,0]==1,:]
        X_val=val_data[:,1]
        T_val=val_data[:,2]
        Y_val=val_data[:,3]
        w = val_data[:,4]
        

        m0, m1 = ICM_matern32(X_E=train_data[train_data[:,0]==1,1], X_O=train_data[train_data[:,0]==0,1], 
                     T_E=train_data[train_data[:,0]==1,2], T_O=train_data[train_data[:,0]==0,2], 
                     Y_E=train_data[train_data[:,0]==1,3], Y_O=train_data[train_data[:,0]==0,3],  
                     r=2, ID=1, AD=0, rho=rho)

        predY0_exp = m0.predict_noiseless(np.c_[np.vstack(val_data[:,1]), np.ones(val_data[:,1].shape[0]) * 0])
        predY1_exp = m1.predict_noiseless(np.c_[np.vstack(val_data[:,1]), np.ones(val_data[:,1].shape[0]) * 0])
        predCATE_exp = predY1_exp[0] - predY0_exp[0]
        varCATE_exp = predY1_exp[1] + predY0_exp[1]

        wmse = weighted_mean_squared_error(weight=w, t = T_val, y_true=Y_val, y_pred0=predY0_exp[0], y_pred1=predY1_exp[0])
        avg_wmse += wmse
        wmse_per_fold.append(wmse)

        print(f"Fold {fold_num + 1} WMSE: {wmse}")
        fold_num += 1
    
    avg_wmse /= num_folds
    print(f"Average WMSE for Rho {rho}: {avg_wmse}")
    wmse_per_rho[rho] = wmse_per_fold

    # Update min_avg_wmse and best_rho if current avg_wmse is lower
    if avg_wmse < min_avg_wmse:
        min_avg_wmse = avg_wmse
        best_rho_matern52 = rho

# Print the best rho and its corresponding min_avg_wmse
print(f"Best Rho: {best_rho_matern52}, Minimum Average WMSE: {min_avg_wmse}")
rmse_matern52 = np.zeros((100))
bias_matern52 = np.zeros((100))
var_matern52 = np.zeros((100))


for i in range(100):
    print(i + 1, end=" ")
    data_exp = datasets[(datasets['S'] == 1) & (datasets['datasetNo'] == i + 1)]
    data_obs = datasets[(datasets['S'] == 0) & (datasets['datasetNo'] == i + 1)]


    m0, m1 = ICM_matern52(X_E = data_exp["X"], X_O = data_obs["X"], 
                 T_E = data_exp["A"], T_O = data_obs["A"], 
                 Y_E = data_exp["Y"], Y_O = data_obs["Y"],  
                 r = 2, ID=1, AD=0, rho=best_rho_matern52)
    predY0 = m0.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predY1 = m1.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predCATE = predY1[0] - predY0[0]

    

    rmse_matern52[i] = np.sqrt(mean_squared_error((predCATE), (trueCATE)))
    bias_matern52[i] = np.mean(predCATE - trueCATE)
    var_matern52[i] = np.mean((predCATE - np.mean(predCATE))**2)


## Polynomial kernel


def ICM_poly(X_E, X_O, T_E, T_O, Y_E, Y_O, r, ID, AD, rho):
    """
    Train and return two rank-2 ICMs
    Args:
    - X_E (numpy.ndarray): Baseline covariate in the experimental data.
    - X_O (numpy.ndarray): Baseline covariate in the observational data.
    - T_E (numpy.ndarray): Treatment indicator in the experimental data.
    - T_O (numpy.ndarray): Treatment indicator in the observational data.
    - Y_E (numpy.ndarray): Outcome in the experimental data.
    - Y_O (numpy.ndarray): Outcome in the observational data.
    - r (int): Rank of the model.
    - ID (int): input dim
    - AD (list): active dim
    - rho: trust hyperparameter

    Returns:
    GPy.models.GPRegression: Trained Gaussian Process Regression model using Intrinsic Coregionalized Model (ICM).

    Data Processing:
    - Split the variables into treated and control for both observational and experimental data.

    Model Training:
    - Intrinsic Coregionalized Model (ICM) with a Radial Basis Function (RBF) kernel is used for regression.
    - The model is trained using Gaussian Process Regression.

    Note:
    - The order of arranging data matters: Experimental control, Observational control, Experimental treated, Observational treated.
    - An indicator is added to the baseline covariate to represent the task (0 for experimental control, 1 for observational control, 2 for experimental treated, 3 for observational treated).
    """

    # Data preprocessing
    # Split the variables in treated and control 
    # Observational Data
    X0_obs = X_O[T_O==0] # baseline covariate in th control arm of the observational study
    X1_obs = X_O[T_O==1] # baseline covariate in th treatment arm of the observational study
    Y0_obs = Y_O[T_O==0] # outcomee in th control arm of the observational study
    Y1_obs = Y_O[T_O==1] # outcomee in th treatment arm of the observational study
    
    #Exoerimental data
    X0_exp = X_E[T_E==0] # baseline covariate in th control arm of the experimental study
    X1_exp = X_E[T_E==1] # baseline covariate in th treatment arm of the experimental study
    Y0_exp = Y_E[T_E==0] # outcomee in th control arm of the experimental study
    Y1_exp = Y_E[T_E==1] # outcomee in th treatment arm of the experimental study
    

    # The order we arrange the data matters
    # 1. Experimental control
    # 2. Observational control
    # 3. Experimental treated
    # 4. Observational treated
    # We also need to add an indicator to the baseline covariate indicating the task
    # i.e. 0 is for the experiemental control, 1 is for the observational control,
    # 2 is for the experiemental treated, 3 is for observational treated
    X0_exp = np.c_[X0_exp,np.ones(X0_exp.shape[0])*0] 
    X0_obs = np.c_[X0_obs,np.ones(X0_obs.shape[0])*1]
    X1_exp = np.c_[X1_exp,np.ones(X1_exp.shape[0])*0]
    X1_obs = np.c_[X1_obs,np.ones(X1_obs.shape[0])*1]

    # Put all the data on a single array 
    Y0 = np.r_[Y0_exp, Y0_obs]
    X0 = np.r_[X0_exp, X0_obs]
    Y1 = np.r_[Y1_exp, Y1_obs]
    X1 = np.r_[X1_exp, X1_obs]


    # model training
    
    # Define the kernel
    kern = GPy.kern.Poly(input_dim=ID, active_dims = AD)**GPy.kern.Coregionalize(input_dim = 1, output_dim = 2, rank=r)
    # Define the model
    m0 = GPy.models.GPRegression(X0,np.vstack(Y0),kern)
    ##############################################################
    # Put all your constraines here
    m0.mul.coregion.W[0,0:1].fix(1)
    m0.mul.coregion.W[0,1:].fix(0)
    m0.mul.coregion.W[1,0:1].fix(rho)
    m0.mul.coregion.W[1,1:].fix(np.sqrt(1-rho**2))
    m0.mul.coregion.kappa.fix([0.000000001, 0.000000001])
    #m0.mul.rbf.variance.constrain_bounded(0., 10.)
    
    ##############################################################
    

    # Optimize the model
    m0.optimize()

    # Define the kernel
    kern = GPy.kern.Poly(input_dim=ID, active_dims = AD)**GPy.kern.Coregionalize(input_dim = 1, output_dim = 2, rank=r)
    # Define the model
    m1 = GPy.models.GPRegression(X1,np.vstack(Y1),kern)
    ##############################################################
    # Put all your constraines here
    m1.mul.coregion.W[0,0:1].fix(1)
    m1.mul.coregion.W[0,1:].fix(0)
    m1.mul.coregion.W[1,0:1].fix(rho)
    m1.mul.coregion.W[1,1:].fix(np.sqrt(1-rho**2))
    m1.mul.coregion.kappa.fix([0.000000001, 0.000000001])
    #m1.mul.rbf.variance.constrain_bounded(0., 10.)

    ##############################################################
    # Optimize the model
    
    m1.optimize()

    return m0, m1

data_exp, data_obs = data_simulation8(1000, beta0=-3.0, beta1=-3.0)
data_full = np.r_[data_exp, data_obs]

logr = linear_model.LogisticRegression()
logr.fit(data_full[:,1].reshape(-1,1),data_full[:,0])
prob = logr.predict_proba(data_full[:,1].reshape(-1,1))
prob_part = prob[:,1]
prob_nonpart = prob[:,0]
data_full = np.c_[data_full, 1/prob_part]

seed_value = 1234

# Define range of rho values to search over
rho_values = np.round(np.arange(0, 1.05, 0.1), 1)

# Define number of folds for cross-validation
num_folds = 5  # Example: 5-fold cross-validation

# Initialize variables to store results
cv_rmse_values = []
rho_rmse_values = {}  # Dictionary to store RMSE values for each rho
min_avg_wmse = float('inf')
best_rho_poly = None
wmse_per_rho = {}

# Perform 5-fold cross-validation
kf = KFold(n_splits=num_folds, shuffle=True)
for rho in rho_values:
    
    print(f"Rho: {rho}")
    avg_wmse = 0
    fold_num = 0
    wmse_per_fold = []
    
    for train_index, val_index in kf.split(data_full):
        # 5 fold CV for experimental data: split them into 5 folds
        train_data, val_data= data_full[train_index], data_full[val_index]
        val_data = val_data[val_data[:,0]==1,:]
        X_val=val_data[:,1]
        T_val=val_data[:,2]
        Y_val=val_data[:,3]
        w = val_data[:,4]
        

        m0, m1 = ICM_matern32(X_E=train_data[train_data[:,0]==1,1], X_O=train_data[train_data[:,0]==0,1], 
                     T_E=train_data[train_data[:,0]==1,2], T_O=train_data[train_data[:,0]==0,2], 
                     Y_E=train_data[train_data[:,0]==1,3], Y_O=train_data[train_data[:,0]==0,3],  
                     r=2, ID=1, AD=0, rho=rho)

        predY0_exp = m0.predict_noiseless(np.c_[np.vstack(val_data[:,1]), np.ones(val_data[:,1].shape[0]) * 0])
        predY1_exp = m1.predict_noiseless(np.c_[np.vstack(val_data[:,1]), np.ones(val_data[:,1].shape[0]) * 0])
        predCATE_exp = predY1_exp[0] - predY0_exp[0]
        varCATE_exp = predY1_exp[1] + predY0_exp[1]

        wmse = weighted_mean_squared_error(weight=w, t = T_val, y_true=Y_val, y_pred0=predY0_exp[0], y_pred1=predY1_exp[0])
        avg_wmse += wmse
        wmse_per_fold.append(wmse)

        print(f"Fold {fold_num + 1} WMSE: {wmse}")
        fold_num += 1
    
    avg_wmse /= num_folds
    print(f"Average WMSE for Rho {rho}: {avg_wmse}")
    wmse_per_rho[rho] = wmse_per_fold

    # Update min_avg_wmse and best_rho if current avg_wmse is lower
    if avg_wmse < min_avg_wmse:
        min_avg_wmse = avg_wmse
        best_rho_poly = rho


# Print the best rho and its corresponding min_avg_wmse
print(f"Best Rho: {best_rho_poly}, Minimum Average WMSE: {min_avg_wmse}")
rmse_poly = np.zeros((100))
bias_poly = np.zeros((100))
var_poly = np.zeros((100))


for i in range(100):
    print(i + 1, end=" ")
    data_exp = datasets[(datasets['S'] == 1) & (datasets['datasetNo'] == i + 1)]
    data_obs = datasets[(datasets['S'] == 0) & (datasets['datasetNo'] == i + 1)]


    m0, m1 = ICM_poly(X_E = data_exp["X"], X_O = data_obs["X"], 
                 T_E = data_exp["A"], T_O = data_obs["A"], 
                 Y_E = data_exp["Y"], Y_O = data_obs["Y"],  
                 r = 2, ID=1, AD=0, rho=best_rho_poly)
    predY0 = m0.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predY1 = m1.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predCATE = predY1[0] - predY0[0]

    

    rmse_poly[i] = np.sqrt(mean_squared_error((predCATE), (trueCATE)))
    bias_poly[i] = np.mean(predCATE - trueCATE)
    var_poly[i] = np.mean((predCATE - np.mean(predCATE))**2)


## Linear Kernel

def ICM_linear(X_E, X_O, T_E, T_O, Y_E, Y_O, r, ID, AD, rho):
    """
    Train and return two rank-2 ICMs
    Args:
    - X_E (numpy.ndarray): Baseline covariate in the experimental data.
    - X_O (numpy.ndarray): Baseline covariate in the observational data.
    - T_E (numpy.ndarray): Treatment indicator in the experimental data.
    - T_O (numpy.ndarray): Treatment indicator in the observational data.
    - Y_E (numpy.ndarray): Outcome in the experimental data.
    - Y_O (numpy.ndarray): Outcome in the observational data.
    - r (int): Rank of the model.
    - ID (int): input dim
    - AD (list): active dim
    - rho: trust hyperparameter

    Returns:
    GPy.models.GPRegression: Trained Gaussian Process Regression model using Intrinsic Coregionalized Model (ICM).

    Data Processing:
    - Split the variables into treated and control for both observational and experimental data.

    Model Training:
    - Intrinsic Coregionalized Model (ICM) with a Radial Basis Function (RBF) kernel is used for regression.
    - The model is trained using Gaussian Process Regression.

    Note:
    - The order of arranging data matters: Experimental control, Observational control, Experimental treated, Observational treated.
    - An indicator is added to the baseline covariate to represent the task (0 for experimental control, 1 for observational control, 2 for experimental treated, 3 for observational treated).
    """

    # Data preprocessing
    # Split the variables in treated and control 
    # Observational Data
    X0_obs = X_O[T_O==0] # baseline covariate in th control arm of the observational study
    X1_obs = X_O[T_O==1] # baseline covariate in th treatment arm of the observational study
    Y0_obs = Y_O[T_O==0] # outcomee in th control arm of the observational study
    Y1_obs = Y_O[T_O==1] # outcomee in th treatment arm of the observational study
    
    #Exoerimental data
    X0_exp = X_E[T_E==0] # baseline covariate in th control arm of the experimental study
    X1_exp = X_E[T_E==1] # baseline covariate in th treatment arm of the experimental study
    Y0_exp = Y_E[T_E==0] # outcomee in th control arm of the experimental study
    Y1_exp = Y_E[T_E==1] # outcomee in th treatment arm of the experimental study
    

    # The order we arrange the data matters
    # 1. Experimental control
    # 2. Observational control
    # 3. Experimental treated
    # 4. Observational treated
    # We also need to add an indicator to the baseline covariate indicating the task
    # i.e. 0 is for the experiemental control, 1 is for the observational control,
    # 2 is for the experiemental treated, 3 is for observational treated
    X0_exp = np.c_[X0_exp,np.ones(X0_exp.shape[0])*0] 
    X0_obs = np.c_[X0_obs,np.ones(X0_obs.shape[0])*1]
    X1_exp = np.c_[X1_exp,np.ones(X1_exp.shape[0])*0]
    X1_obs = np.c_[X1_obs,np.ones(X1_obs.shape[0])*1]

    # Put all the data on a single array 
    Y0 = np.r_[Y0_exp, Y0_obs]
    X0 = np.r_[X0_exp, X0_obs]
    Y1 = np.r_[Y1_exp, Y1_obs]
    X1 = np.r_[X1_exp, X1_obs]


    # model training
    
    # Define the kernel
    kern = GPy.kern.Linear(input_dim=ID, active_dims = AD)**GPy.kern.Coregionalize(input_dim = 1, output_dim = 2, rank=r)
    # Define the model
    m0 = GPy.models.GPRegression(X0,np.vstack(Y0),kern)
    ##############################################################
    # Put all your constraines here
    m0.mul.coregion.W[0,0:1].fix(1)
    m0.mul.coregion.W[0,1:].fix(0)
    m0.mul.coregion.W[1,0:1].fix(rho)
    m0.mul.coregion.W[1,1:].fix(np.sqrt(1-rho**2))
    m0.mul.coregion.kappa.fix([0.000000001, 0.000000001])
    #m0.mul.rbf.variance.constrain_bounded(0., 10.)
    
    ##############################################################
    

    # Optimize the model
    m0.optimize()

    # Define the kernel
    kern = GPy.kern.Linear(input_dim=ID, active_dims = AD)**GPy.kern.Coregionalize(input_dim = 1, output_dim = 2, rank=r)
    # Define the model
    m1 = GPy.models.GPRegression(X1,np.vstack(Y1),kern)
    ##############################################################
    # Put all your constraines here
    m1.mul.coregion.W[0,0:1].fix(1)
    m1.mul.coregion.W[0,1:].fix(0)
    m1.mul.coregion.W[1,0:1].fix(rho)
    m1.mul.coregion.W[1,1:].fix(np.sqrt(1-rho**2))
    m1.mul.coregion.kappa.fix([0.000000001, 0.000000001])
    #m1.mul.rbf.variance.constrain_bounded(0., 10.)

    ##############################################################
    # Optimize the model
    
    m1.optimize()

    return m0, m1



data_exp, data_obs = data_simulation8(1000, beta0=-3.0, beta1=-3.0)
data_full = np.r_[data_exp, data_obs]

logr = linear_model.LogisticRegression()
logr.fit(data_full[:,1].reshape(-1,1),data_full[:,0])
prob = logr.predict_proba(data_full[:,1].reshape(-1,1))
prob_part = prob[:,1]
prob_nonpart = prob[:,0]
data_full = np.c_[data_full, 1/prob_part]

seed_value = 1234

# Define range of rho values to search over
rho_values = np.round(np.arange(0, 1.05, 0.1), 1)

# Define number of folds for cross-validation
num_folds = 5  # Example: 5-fold cross-validation

# Initialize variables to store results
cv_rmse_values = []
rho_rmse_values = {}  # Dictionary to store RMSE values for each rho
min_avg_wmse = float('inf')
best_rho_linear = None
wmse_per_rho = {}

# Perform 5-fold cross-validation
kf = KFold(n_splits=num_folds, shuffle=True)
for rho in rho_values:
    
    print(f"Rho: {rho}")
    avg_wmse = 0
    fold_num = 0
    wmse_per_fold = []
    
    for train_index, val_index in kf.split(data_full):
        # 5 fold CV for experimental data: split them into 5 folds
        train_data, val_data= data_full[train_index], data_full[val_index]
        val_data = val_data[val_data[:,0]==1,:]
        X_val=val_data[:,1]
        T_val=val_data[:,2]
        Y_val=val_data[:,3]
        w = val_data[:,4]
        

        m0, m1 = ICM_matern32(X_E=train_data[train_data[:,0]==1,1], X_O=train_data[train_data[:,0]==0,1], 
                     T_E=train_data[train_data[:,0]==1,2], T_O=train_data[train_data[:,0]==0,2], 
                     Y_E=train_data[train_data[:,0]==1,3], Y_O=train_data[train_data[:,0]==0,3],  
                     r=2, ID=1, AD=0, rho=rho)

        predY0_exp = m0.predict_noiseless(np.c_[np.vstack(val_data[:,1]), np.ones(val_data[:,1].shape[0]) * 0])
        predY1_exp = m1.predict_noiseless(np.c_[np.vstack(val_data[:,1]), np.ones(val_data[:,1].shape[0]) * 0])
        predCATE_exp = predY1_exp[0] - predY0_exp[0]
        varCATE_exp = predY1_exp[1] + predY0_exp[1]

        wmse = weighted_mean_squared_error(weight=w, t = T_val, y_true=Y_val, y_pred0=predY0_exp[0], y_pred1=predY1_exp[0])
        avg_wmse += wmse
        wmse_per_fold.append(wmse)

        print(f"Fold {fold_num + 1} WMSE: {wmse}")
        fold_num += 1
    
    avg_wmse /= num_folds
    print(f"Average WMSE for Rho {rho}: {avg_wmse}")
    wmse_per_rho[rho] = wmse_per_fold

    # Update min_avg_wmse and best_rho if current avg_wmse is lower
    if avg_wmse < min_avg_wmse:
        min_avg_wmse = avg_wmse
        best_rho_linear = rho

# Print the best rho and its corresponding min_avg_wmse
print(f"Best Rho: {best_rho_linear}, Minimum Average WMSE: {min_avg_wmse}")
rmse_linear = np.zeros((100))
bias_linear = np.zeros((100))
var_linear = np.zeros((100))


for i in range(100):
    print(i + 1, end=" ")
    data_exp = datasets[(datasets['S'] == 1) & (datasets['datasetNo'] == i + 1)]
    data_obs = datasets[(datasets['S'] == 0) & (datasets['datasetNo'] == i + 1)]


    m0, m1 = ICM_linear(X_E = data_exp["X"], X_O = data_obs["X"], 
                 T_E = data_exp["A"], T_O = data_obs["A"], 
                 Y_E = data_exp["Y"], Y_O = data_obs["Y"],  
                 r = 2, ID=1, AD=0, rho=best_rho_linear)
    predY0 = m0.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predY1 = m1.predict_noiseless(np.c_[np.vstack(test_data),np.ones(test_data.shape[0])*0])
    predCATE = predY1[0] - predY0[0]

    

    rmse_linear[i] = np.sqrt(mean_squared_error((predCATE), (trueCATE)))
    bias_linear[i] = np.mean(predCATE - trueCATE)
    var_linear[i] = np.mean((predCATE - np.mean(predCATE))**2)


print(rho_RBF)
print(np.round(np.mean(rmse_RBF), 3))
print(np.round(np.std(rmse_RBF), 3))
print('\n')

print(best_rho_matern32)
print(np.round(np.mean(rmse_matern32), 3))
print(np.round(np.std(rmse_matern32), 3))
print('\n')

print(best_rho_matern52)
print(np.round(np.mean(rmse_matern52), 3))
print(np.round(np.std(rmse_matern52), 3))
print('\n')

print(best_rho_poly)
print(np.round(np.mean(rmse_poly), 3))
print(np.round(np.std(rmse_poly), 3))
print('\n')

print(best_rho_linear)
print(np.round(np.mean(rmse_linear), 3))
print(np.round(np.std(rmse_linear), 3))


# Save the results
sens_RBF = pd.DataFrame({"RMSE": rmse_RBF, "Bias": bias_RBF, "Variance": var_RBF})
sens_matern32 = pd.DataFrame({"RMSE": rmse_matern32, "Bias": bias_matern32, "Variance": var_matern32})
sens_matern52 = pd.DataFrame({"RMSE": rmse_matern52, "Bias": bias_matern52, "Variance": var_matern52})
sens_poly = pd.DataFrame({"RMSE": rmse_poly, "Bias": bias_poly, "Variance": var_poly})
sens_linear = pd.DataFrame({"RMSE": rmse_poly, "Bias": bias_poly, "Variance": var_poly})

# Create a list of the DataFrames
dfs = [sens_RBF, sens_matern32, sens_matern52, sens_poly, sens_linear]

# Create a list of overlap values
kernel_values = ['RBF', 'Matérn 3/2', 'Matérn 5/2', 'Polynomial', 'Linear']

# Add the 'overlap' column to each DataFrame
for df, kernel in zip(dfs, kernel_values):
    df['Kernel'] = kernel

# Concatenate all the DataFrames into one
sens_kernel = pd.concat(dfs, ignore_index=True)

# Save the combined DataFrame to a CSV file
sens_kernel.to_csv('sens_kernel.csv', index=False)

# Display the combined DataFrame
print(sens_kernel)
```
